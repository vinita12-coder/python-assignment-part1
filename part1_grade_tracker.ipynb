# ***Task 1:- Data Parsing & Profile Cleaning***

```python
raw_students = [
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma",       "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ",    "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA",       "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ",    "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

cleaned_students = []  # will store all cleaned student dicts

for student in raw_students:

    #1: Clean the name
    clean_name = student["name"].strip().title()
    ## strip()- Removes the extra spaces
    ## title()- Convert first letter capital

    #2:Converting Roll no from sring to int
    clean_roll = int(student["roll"])

    #3: marks_str convert in to int
    clean_marks = [int(m) for m in student["marks_str"].split(", ")]

    #4: Validate name
    words = clean_name.split()
    is_valid = all(word.isalpha() for word in words)

    if is_valid:
        print(f"{clean_name} ✓ Valid name")
    else:
        print(f"{clean_name} ✗ Invalid name")

    #5: Print the profile card
    print("================================")
    print(f"Student : {clean_name}")
    print(f"Roll No : {clean_roll}")
    print(f"Marks   : {clean_marks}")
    print("================================")

    # Saving clean data into new dictionary
    cleaned_students.append({
        "name": clean_name,
        "roll": clean_roll,
        "marks": clean_marks
    })

#6: Finding roll 103 and print ALL CAPS and lowercase
for student in cleaned_students:
    if student["roll"] == 103:
        print(student["name"].upper())   # ALL CAPS
        print(student["name"].lower())   # all lowercase
```

# ***Task 2 — Marks Analysis Using Loops & Conditionals***

```python
#Part 1- For Loops
student_name = "Ayesha Sharma"
subjects     = ["Math", "Physics", "CS", "English", "Chemistry"]
marks        = [88, 72, 95, 60, 78]

# --- Part 1: Print each subject with grade using for loop ---

print(f"Results for {student_name}:\n")

for i in range(len(subjects)):
    m = marks[i]

    # Assign grade based on marks range
    if m >= 90:
        grade = "A+"
    elif m >= 80:
        grade = "A"
    elif m >= 70:
        grade = "B"
    elif m >= 60:
        grade = "C"
    else:
        grade = "F"

    print(f"{subjects[i]}: {m} — {grade}")

# --- Part 2: Stats ---

total   = sum(marks)
average = round(total / len(marks), 2)

# zip pairs subjects with marks so we can find max/min together
highest_sub = subjects[marks.index(max(marks))]
lowest_sub  = subjects[marks.index(min(marks))]

print(f"\nTotal   : {total}")
print(f"Average : {average}")
print(f"Highest : {highest_sub} ({max(marks)})")
print(f"Lowest  : {lowest_sub} ({min(marks)})")

# --- Part 3: while loop — adding new subjects ---

new_count = 0
while True:
    sub_name = input("\nEnter subject name (or 'done' to stop): ")

    if sub_name.lower() == "done":
        break

    mark_input = input(f"Enter marks for {sub_name} (0–100): ")

    # Validate: must be a number and in range 0-100
    if not mark_input.isnumeric():
        print(" Invalid input: not a number. Skipping.")
        continue

    mark_val = int(mark_input)

    if mark_val < 0 or mark_val > 100:
        print("Invalid input: marks must be between 0 and 100. Skipping.")
        continue

    # Valid — add to lists
    subjects.append(sub_name)
    marks.append(mark_val)
    new_count += 1

# After loop ends
updated_avg = round(sum(marks) / len(marks), 2)

print(f"\nNew subjects added : {new_count}")
print(f"Updated average    : {updated_avg}")
```

# ***Task 3 — Class Performance Summary***

```python
class_data = [
    ("Ayesha Sharma",  [88, 72, 95, 60, 78]),
    ("Rohit Verma",    [55, 68, 49, 72, 61]),
    ("Priya Nair",     [91, 85, 88, 94, 79]),
    ("Karan Mehta",    [40, 55, 38, 62, 50]),
    ("Sneha Pillai",   [75, 80, 70, 68, 85]),
]

# --- Printing table header ---
print(f"{'Name':<18}| {'Average':<7} | Status")
print("-" * 40)

all_averages = []

for name, marks in class_data:

    # Calculateing average rounded to 2 decimal places
    avg = round(sum(marks) / len(marks), 2)

    # Pass if average >= 60, else Fail
    status = "Pass" if avg >= 60 else "Fail"

    # Printing formatted row
    print(f"{name:<18}| {avg:^7} | {status}")

    # Store for later use
    all_averages.append((name, avg, status))

# --- Stats after table ---

passed = [s for s in all_averages if s[2] == "Pass"]
failed = [s for s in all_averages if s[2] == "Fail"]

print(f"\nPassed : {len(passed)}")
print(f"Failed : {len(failed)}")

# Topper = student with highest average score
topper = max(all_averages, key=lambda x: x[1])
print(f"Topper : {topper[0]} ({topper[1]})")

# Class average = average of each averages
class_avg = round(sum(a[1] for a in all_averages) / len(all_averages), 2)
print(f"Class Average : {class_avg}")
```

# ***Task 4 — String Manipulation Utility***

```python
essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "

# Step 1: Strip extra space
clean_essay = essay.strip()
print("Step 1 - Stripped:")
print(clean_essay)

# Step 2: Conversion
print("\nStep 2 - Title Case:")
print(clean_essay.title())

# Step 3: Count "python" (case-insensitive)
python_count = clean_essay.count("python")
print(f"\nStep 3 - 'python' appears {python_count} time(s)")

# Step 4: Replace "python" with "Python 🐍"
replaced = clean_essay.replace("python", "Python 🐍")
print("\nStep 4 - Replaced:")
print(replaced)

# Step 5: Split into sentences by ". "
sentences = clean_essay.split(". ")
print("\nStep 5 - Sentences list:")
print(sentences)

# Step 6: Print each sentence numbered, add "." if not already there
print("\nStep 6 - Numbered sentences:")
for i, sentence in enumerate(sentences, start=1):
    if not sentence.endswith("."):
        sentence = sentence + "."
    print(f"{i}. {sentence}")
```
